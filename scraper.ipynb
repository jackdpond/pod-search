```python
def clean_name(url):
    name = url.split('/')[-1]
    return name.replace('-', ' ').title()

clean_name('https://bibleproject.com/podcast/how-did-new-testament-come-be')
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import os

options = webdriver.ChromeOptions()
options.add_argument("--headless")
driver = webdriver.Chrome(options=options)

driver = webdriver.Chrome()
driver.get("https://bibleproject.com/podcasts/the-bible-project-podcast/?tab=series")

os.makedirs('Bible_Project')

series_links = series_links = set(el.get_attribute("href") for el in driver.find_elements(By.CSS_SELECTOR, 'a[href^="/podcast/series/"]'))

for url in series_links:
    if url:
        print(f"Entering series {clean_name(url)}")
        driver.get(url)  # <-- replace with the actual URL

        time.sleep(1)

        episode_links = driver.find_elements(By.CSS_SELECTOR, 'a[href^="/podcast/"]')
        unique_links = set(link.get_attribute("href") for link in episode_links)

        for l in unique_links:
            if l:
                print(f"Entering episode {clean_name(l)}")
                driver.get(l)
                time.sleep(1)

                try:
                    mp3_url = driver.find_element(By.CSS_SELECTOR, 'a[download]').get_attribute("href")
                    print(f"download link: {mp3_url}")
                except Exception as e:
                    mp3_url = None
            else:
                print("no episode link found")
    else:
        print("no series link found")
```
